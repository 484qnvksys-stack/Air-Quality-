**Class Distribution:** From our class label (air quality) there are four different classes: Good, Moderate, Poor, and Hazardous.

```python
import pandas as pd

df = pd.read_csv("Raw_dataset.csv")

print("Target Column (class attribute): Air Quality")

count_good = df[df["Air Quality"] == "Good"].shape[0]
count_moderate = df[df["Air Quality"] == "Moderate"].shape[0]
count_poor = df[df["Air Quality"] == "Poor"].shape[0]
count_hazardous = df[df["Air Quality"] == "Hazardous"].shape[0]

print(f"- Instances with Good air quality: {count_good}")
print(f"- Instances with Moderate air quality: {count_moderate}")
print(f"- Instances with Poor air quality: {count_poor}")
print(f"- Instances with Hazardous air quality: {count_hazardous}")
```
